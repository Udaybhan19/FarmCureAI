```python
import os
import shutil
import numpy as np
from PIL import Image

# Paths for NLB dataset
nlb_root_dir = "/content/drive/MyDrive/Uday/NLB_Dataset/Total_Dataset"

nlb_train_images = os.path.join(nlb_root_dir, "train_images")
nlb_train_masks = os.path.join(nlb_root_dir, "train_semantic")

nlb_test_images = os.path.join(nlb_root_dir, "test_images")
nlb_test_masks = "/content/drive/MyDrive/Uday/Prediction_Images_Array/NLB_InternImage/gt_masks_corrected"

# Destination
dest_root_dir = "/content/drive/MyDrive/Uday/All_Disease_Dataset/Cropwise_Disease_Dataset/corn"

dest_train_images = os.path.join(dest_root_dir, "train", "images")
dest_train_masks = os.path.join(dest_root_dir, "train", "masks_label")

dest_test_images = os.path.join(dest_root_dir, "test", "images")
dest_test_masks = os.path.join(dest_root_dir, "test", "masks_label")

# Ensure destination directories exist
for d in [
    dest_train_images,
    dest_train_masks,
    dest_test_images,
    dest_test_masks
]:
    os.makedirs(d, exist_ok=True)


def copy_with_prefix(
    src_images_dir,
    src_masks_dir,
    dest_images_dir,
    dest_masks_dir,
    prefix
):

    image_files = [
        f for f in os.listdir(src_images_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]

    for img_file in image_files:

        base_name, ext = os.path.splitext(img_file)

        new_name = f"{prefix}_{base_name}{ext}"
        new_mask_name = f"{prefix}_{base_name}.png"

        src_image_path = os.path.join(src_images_dir, img_file)
        src_mask_path = os.path.join(src_masks_dir, base_name + ".png")

        dest_image_path = os.path.join(dest_images_dir, new_name)
        dest_mask_path = os.path.join(dest_masks_dir, new_mask_name)


        mask = np.array(Image.open(src_mask_path))


        original_unique = np.unique(mask)

        # -----------------------------------
        # Convert 1 -> 2
        # -----------------------------------
        mask[mask == 1] = 2

        # -----------------------------------
        # Check converted unique pixels
        # -----------------------------------
        converted_unique = np.unique(mask)

        # Check whether mask contains ONLY 0 and 2
        if not np.all(np.isin(converted_unique, [0, 2])):
            print(
                f"⚠️ WARNING: Unexpected pixel values in {img_file}"
            )
            print("Original unique pixels:", original_unique)
            print("Converted unique pixels:", converted_unique)

            # Skip this mask/image
            continue

        # Print verification
        print(
            f"✅ {img_file} | "
            f"Original: {original_unique} | "
            f"Converted: {converted_unique}"
        )

        # -----------------------------------
        # Copy image
        # -----------------------------------
        shutil.copy2(
            src_image_path,
            dest_image_path
        )

        # -----------------------------------
        # Save converted mask
        # -----------------------------------
        Image.fromarray(
            mask.astype(np.uint8)
        ).save(dest_mask_path)


# Copy Train set
copy_with_prefix(
    nlb_train_images,
    nlb_train_masks,
    dest_train_images,
    dest_train_masks,
    "NLB_Blight_Disease"
)

# Copy Test set
copy_with_prefix(
    nlb_test_images,
    nlb_test_masks,
    dest_test_images,
    dest_test_masks,
    "NLB_Blight_Disease"
)

print("✅ Dataset copied and masks converted successfully!")
```

```python
from pathlib import Path

DATASET_PATH = Path("/content/drive/MyDrive/Uday/All_Disease_Dataset/Cropwise_Disease_Dataset")

crops = [
    "apple",
    "grape",
    "rice",
    "wheat",
    "soybean",
    "corn",
    "tomato",
    "banana",
    "potato",
    "cabbage",
    "bean",
    "cucumber"
]

# Common image extensions
valid_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp"}


def count_images(folder: Path) -> int:
    """Count image files in a folder."""
    if not folder.exists():
        return 0

    return sum(
        1
        for file in folder.iterdir()
        if file.is_file() and file.suffix.lower() in valid_extensions
    )


print(
    f"{'Crop':<12} "
    f"{'Train Images':>13} "
    f"{'Train Masks':>13} "
    f"{'Test Images':>12} "
    f"{'Test Masks':>12} "
    f"{'Total Images':>13}"
)

print("-" * 82)

for crop in crops:
    crop_path = DATASET_PATH / crop

    if not crop_path.exists():
        print(f"{crop:<12} {'FOLDER NOT FOUND':>30}")
        continue

    train_images = count_images(crop_path / "train" / "images")
    train_masks = count_images(crop_path / "train" / "masks_label")

    test_images = count_images(crop_path / "test" / "images")
    test_masks = count_images(crop_path / "test" / "masks_label")

    total_images = train_images + test_images

    print(
        f"{crop:<12} "
        f"{train_images:>13} "
        f"{train_masks:>13} "
        f"{test_images:>12} "
        f"{test_masks:>12} "
        f"{total_images:>13}"
    )
```
